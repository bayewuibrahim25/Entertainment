```python
# Rename the CO2 emissions column to a clean name
df = df.rename(columns={"CO__emissions_per_capita": "co2_per_capita"})
df.columns
```

```python
# Save cleaned DataFrame to SQLite
df.to_sql("co2_data", conn, if_exists="replace", index=False)
```

```python
# Top 10 emitters in the latest year
query1 = """
SELECT Entity, Year, co2_per_capita
FROM co2_data
WHERE Year = (SELECT MAX(Year) FROM co2_data)
ORDER BY co2_per_capita DESC
LIMIT 10;
"""
top_emitters = pd.read_sql(query1, conn)
top_emitters
```

Query 1: Top 10 Emitters

This query finds the top 10 countries with the highest CO₂ emissions per capita in the most recent year. It uses a subquery to automatically select the latest year, sorts the data from highest to lowest emissions, and limits the results to 10 countries. This helps identify the biggest emitters per person. It highlights which countries have the greatest environmental impact per person, allowing for fair comparison between countries regardless of population size.

```python
# Ireland emissions over time
query2 = """
SELECT Year, co2_per_capita
FROM co2_data
WHERE Entity = 'Ireland'
ORDER BY Year;
"""
ireland_data = pd.read_sql(query2, conn)
ireland_data
```

Query 2: Ireland Over Time

This query shows Ireland’s CO₂ emissions per capita over time. It filters the data for Ireland only and orders it by year. This allows us to see trends and changes in emissions across different years. It shows how Ireland’s emissions have changed over time, helping us identify trends such as increases, decreases, or stability.

Interactive Visualization using Widgets

To make the project interactive, an ipywidgets dropdown menu was used to allow users to select a country and view its CO₂ emissions over time.

```python
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt

def show_country(country):
    data = df[df['Entity'] == country]
    plt.plot(data['Year'], data['co2_per_capita'])
    plt.title(f'CO2 Emissions: {country}')
    plt.xlabel('Year')
    plt.ylabel('CO2 per Capita')
    plt.show()

dropdown = widgets.Dropdown(options=df['Entity'].unique())
widgets.interact(show_country, country=dropdown)
```
